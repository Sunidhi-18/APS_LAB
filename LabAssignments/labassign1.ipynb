Outcomes for 20 customers

```python
import numpy as np
p = 0.30
sample_size=20
np.random.seed(42)
samples = np.random.binomial(n=1, p=p, size = sample_size)
```

Count clicks adnd non-Clicks

```python
n_clicks = np.sum(samples == 1)
n_noclicks = np.sum(samples == 0)

print("number of clicks: ",n_clicks)
print("number of non-clicks: ",n_noclicks)
```

```python
import matplotlib.pyplot as plt
outcomes = {0,1}
theoretical_pmf = [0.70, 0.30]
empirical_pmf = [n_noclicks / 20, n_clicks / 20]

x = np.arange(len(outcomes))
width = 0.35

plt.bar(x - width / 2, theoretical_pmf, width, label = 'Theoretical')
plt.bar(x - width / 2, empirical_pmf, width, label = 'Empirical')
plt.xlabel('Outcome (X)')
plt.ylabel('Probability')
plt.xticks(x, ['0 (no click)', '1 (click)'])
plt.legend()
plt.title('Theoretical vs Empirical PMF (N=20)')
plt.show()
```

```python
sample_sizes = [10,100,1000,10000]
for n in sample_sizes:
    s = np.random.binomial(n = 1, p = 0.30, size = n)
    p0 = np.mean(s == 0)
    p1 = np.mean(s == 1)
    mean = np.mean(s)
    var = np.var(s)
    print(f"N={n:5d} | P(X=0): {p0:.4f} | P(X=1): {p1:.4f} | Mean: {mean:.4f} | Var: {var:.4f}")
```
