Simulating operations and Calculating key metrics

```python
import numpy as np

# Set parameters
n_transactions = 50
p_flag = 0.04
simulations = 20000
capacity = 4

np.random.seed(42)  # For reproducible results

# 1. Simulate 20,000 hours of operations
flagged_counts = np.random.binomial(n=n_transactions, p=p_flag, size=simulations)

# Calculate key stats
avg_flagged = np.mean(flagged_counts)
prob_exceed_capacity = np.mean(flagged_counts > capacity)
unexamined_transactions = np.maximum(0, flagged_counts - capacity)
avg_unexamined = np.mean(unexamined_transactions)

print(f"Average flagged transactions per hour: {avg_flagged:.4f}")
print(f"Probability of exceeding capacity (> 4): {prob_exceed_capacity:.4f}")
print(f"Average unexamined flagged transactions per hour: {avg_unexamined:.4f}")
```
