```python
#1: A fair coin is tossed 100 times.Calculate: 
#Number of heads  
#Number of tails 
#Experimental probability of Head  
#Experimental probability of Tail 

import random

n_tosses = 100
#simulate coin tosses by randomly choosing heads or tails
tosses = [random.choice(['heads', 'tails']) for _ in range(n_tosses)]

#occurences of heads and tails
num_heads = tosses.count('heads')
num_tails = tosses.count('tails')

#calculate experimental prob for both outcomes
exp_prob_heads = num_heads / n_tosses
exp_prob_tails = num_tails / n_tosses

print("number of heads: ", {num_heads})
print("number of tails: ", {num_tails})
print("experimental probability of heads: ", {exp_prob_heads})
print("experimental probability of tails: ", {exp_prob_tails})
```

```python
#2: check Effect of the number of trials 
#Repeat the coin-toss experiment for: n=10, 100, 1,000, 10,000 

import random

#loop through different sample sizes to see the effect of trials
for n in[10, 100, 1000, 10000]:
    #simulate coin tosses (1->heads, 0->tails) & count total heads
    heads = sum(random.choice([1,0]) for _ in range(n))

    prob = heads / n
    print(f"n= {n: 5d} -> Heads= {heads: 5d} -> Probability= {prob: .4f}")
```

```python
#3: Rolling a die: A fair die is rolled 6,000 times. 
#Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

import random

# Simulate rolling a fair die 6,000 times
rolls = [random.randint(1,6) for _ in range(6000)]

# Calculate and print the experimental probability for each outcome (1-6)
for i in range(1,7):
    probability = rolls.count(i) / 6000
    print(f"outcome {i}: {probability: .4f}")
```

```python
#4: Addition rule for disjoint events 
#A die is rolled 10,000 times. 
#Define: A=getting 1, B= getting 6, check the events are joint or disjoint  
#Verify: P(A∪B)=P(A)+P(B) 

import random

rolls = [random.randint(1,6) for _ in range(10000)]

# Calculate the experimental probability of getting a 1 and 6
P_A = rolls.count(1) / 10000
P_B = rolls.count(6) / 10000
# Calculate the experimental probability of getting either 1 or 6
P_union= sum(1 for r in rolls if r ==1 or r==6) / 10000

print(f"P(A) + P(B): {P_A + P_B: .4f}")
print(f"P(A U B): {P_union: .4f}")
```

```python
#5. Addition rule for overlapping events 
#For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6} 
#Verify: P(A∪B)=P(A)+P(B)−P(A∩B)

import random
rolls = [random.randint(1,6) for _ in range(10000)]

P_A = sum(1 for r in rolls if r in [2,4,6]) / 10000
P_B = sum(1 for r in rolls if r in [4,5,6]) / 10000
P_intersect = sum(1 for r in rolls if r in [4,6]) / 10000
P_union = sum(1 for r in rolls if r in [2,4,5,6]) / 10000

# Calculate P(A U B) using the general addition rule formula for overlapping events
calc_union = P_A + P_B - P_intersect

print(f"P(A) + P(B) - P(A n B): {calc_union: .4f}")
print(f"Actual union P(A U B): {P_union: .4f}")
```

```python
#6. Multiplication rule for independent events 
#A coin is tossed and a die is rolled 10,000 times. 
#Define: 
#obtaining Head  
#obtaining an even number 
#Verify: P(A∩B)=P(A)P(B) 

import random

trials = 10000
cnt_A = 0
cnt_B = 0
cnt_intersection = 0

# Simulate tossing a coin and rolling a die 10,000 times
for _ in range(trials):
    coin = random.choice(['H','T'])
    die = random.randint(1,6)

    is_A = (coin == 'H')
    is_B = (die in [2,4,6])

    if is_A:
        cnt_A = cnt_A + 1
    if is_B:
        cnt_B = cnt_B + 1
    if is_A and is_B:
        cnt_intersection = cnt_intersection + 1

# Calculate experimental probabilities
P_A = cnt_A / trials
P_B = cnt_B / trials
P_intersect = cnt_intersection / trials
P_product = P_A * P_B

print(f"P(A): {P_A:.4f}")
print(f"P(B): {P_B: .4f}")
print(f"P(A n B): {P_intersect: .4f}")
print(f"P(A)* P(B) [calc]: {P_product: .4f}")

```
