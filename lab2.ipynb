#The Birthday Problem - In a group of people, what is the probability that at least 2 people share the same birthday

```python
n = int(input("enter no.of persons "))
total_days = 365

prob_no_match = 1
for i in range(n):
    prob_no_match = prob_no_match*(total_days-i)/total_days

prob_shared = 1-prob_no_match
print(f"probability of {n} persons: {prob_shared*100:.4f}%")
```

```python
import matplotlib.pyplot as plt

def test_bday(max_val=50):
    total_days = 365
    
    # Test specific inputs
    test_inputs = [3, 10, 15, 20, 23]
    for n in test_inputs:
        prob_no_match = 1
        for i in range(n):
            prob_no_match = prob_no_match * (total_days - i) / total_days
        prob_shared = 1 - prob_no_match
        print(f"probability of {n} persons: {prob_shared * 100:.4f}%")
    print("-" * 31)

    # Prepare data for plotting
    n_values = list(range(1, max_val + 1))
    probabilities = []
    for n in n_values:
        prob_no_match = 1.0
        for i in range(n):
            prob_no_match *= (total_days - i) / total_days
        probabilities.append((1 - prob_no_match) * 100)

    # Plotting the results
    plt.figure(figsize=(10, 6))
    plt.plot(n_values, probabilities, color='#2b5c8f', linewidth=2.5, label='Probability Curve')
    
    # Calculate and mark the test inputs on the plot
    highlight_probs = []
    for n in test_inputs:
        p_no = 1.0
        for i in range(n):
            p_no *= (total_days - i) / total_days
        highlight_probs.append((1 - p_no) * 100)
        
    plt.scatter(test_inputs, highlight_probs, color='#d9534f', zorder=5, s=60)
    for n, p in zip(test_inputs, highlight_probs):
        plt.text(n + 0.8, p - 1.5, f"n={n}\n({p:.1f}%)", fontsize=9, fontweight='bold', color='#d9534f')

    # Graph formatting
    plt.axhline(y=50, color='gray', linestyle='--', alpha=0.7, label='50% Threshold ($n=23$)')
    plt.title('The Birthday Problem: Probability vs. Number of People', fontsize=14, fontweight='bold')
    plt.xlabel('Number of People ($n$)', fontsize=12)
    plt.ylabel('Probability (%)', fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(fontsize=11, loc='lower right')
    plt.tight_layout()
    
    # Save and show graph
    plt.savefig('birthday_probability.png', dpi=300)
    plt.show()

# Run the function
test_bday(50)
```
